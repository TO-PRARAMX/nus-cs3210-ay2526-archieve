# Computation Time

```python
fig_car, ax_car = plt.subplots(figsize=(3.3,2.5))
ax_car.set_xscale('log')
ax_car.errorbar(df_L10n['cars_size'],
                df_L10n['t_exe'],
                yerr=df_L10n['t_exe_err']/100 * df_L10n['t_exe'],
                label=r"$L = 10n$")
ax_car.errorbar(df_L20n['cars_size'],
                df_L20n['t_exe'],
                yerr=df_L20n['t_exe_err']/100 * df_L20n['t_exe'],
                label=r"$L = 20n$")
ax_car.errorbar(df_L50n['cars_size'],
                df_L50n['t_exe'],
                yerr=df_L50n['t_exe_err']/100 * df_L50n['t_exe'],
                label=r"$L = 50n$")
ax_car.errorbar(df_L100n['cars_size'],
                df_L100n['t_exe'],
                yerr=df_L100n['t_exe_err']/100 * df_L100n['t_exe'],
                label=r"$L = 100n$")

ax_car.grid()
ax_car.set_xlabel("array size (N)")
ax_car.set_xticks([1e4,1e5,1e6,1e7])
# ax_car.set_title("Performance scaling on\n the Number of Cars")
ax_car.set_ylabel("execution time (s)")
ax_car.legend(fontsize='8')
fig_car.savefig(fname="Images/number-of-cars.pdf")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df_w5_3423_thread  = pd.read_csv('csv/w5-3423-thread.csv')
df_xs_4114_thread  = pd.read_csv('csv/xs-4114-thread.csv')
df_i7_7700_thread  = pd.read_csv('csv/i7-7700-thread.csv')
df_i7_13700_thread = pd.read_csv('csv/i7-13700-thread.csv')

# i7-7700 doesn't have 16 cores => put 0 (manually) in all columns beside num_threads at the last line

num_threads = df_w5_3423_thread['thread_nums'].to_numpy()

fig_cpu_thread, axs_cpu_thread = plt.subplots(figsize=(4.8, 2.1), ncols=2, sharex=True)

X = np.arange(len(num_threads))
w = 0.2

x_w5     = X - 1.5*w
x_xs     = X - 0.5*w
x_7700   = X + 0.5*w
x_13700  = X + 1.5*w


# execution time vs number of threads
axs_cpu_thread[0].bar(x_w5,    df_w5_3423_thread['execution_time'],  width=w, label="w5-3423")
axs_cpu_thread[0].bar(x_xs,    df_xs_4114_thread['execution_time'],  width=w, label="xs-4114")
axs_cpu_thread[0].bar(x_7700,  df_i7_7700_thread['execution_time'],  width=w, label="i7-7700")
axs_cpu_thread[0].bar(x_13700, df_i7_13700_thread['execution_time'], width=w, label="i7-13700")

# speed-up vs number of threads
axs_cpu_thread[1].bar(x_w5,    df_w5_3423_thread['speed_up'],  width=w, label="w5-3423")
axs_cpu_thread[1].bar(x_xs,    df_xs_4114_thread['speed_up'],  width=w, label="xs-4114")
axs_cpu_thread[1].bar(x_7700,  df_i7_7700_thread['speed_up'],  width=w, label="i7-7700")
axs_cpu_thread[1].bar(x_13700, df_i7_13700_thread['speed_up'], width=w, label="i7-13700")

for ax in axs_cpu_thread:
    ax.grid(axis="y", alpha=0.3)
    ax.legend(fontsize=8)
    ax.set_xticks(X)
    ax.set_xticklabels(num_threads)

axs_cpu_thread[0].set_ylabel("execution time (s)")
axs_cpu_thread[1].set_ylabel("speed up")
fig_cpu_thread.supxlabel("number of threads", y=0.08)

fig_cpu_thread.tight_layout()
fig_cpu_thread.savefig("Images/cpu-threads.pdf")
plt.show()
```
